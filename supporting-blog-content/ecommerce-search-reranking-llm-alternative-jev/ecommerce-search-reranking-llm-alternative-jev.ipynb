## Limitations

- The benchmark evaluates ordering inside complete judged ESCI candidate groups; it does not measure production retrieval recall or online user outcomes.
- The groups are relevance-dense, so random nDCG is high and top-10 nDCG has less headroom than it would in a candidate set with more irrelevant products.
- ESCI validates the four-way relationship, not the auxiliary Noul signals used by the composite policy. Those signals require independent human labels.
- The instruction variants were selected automatically by a checked-in, preregistered rule on a separate development run, then evaluated on the reserved confirmation sample. This supports the controlled comparison but does not substitute for replication on another dataset or online outcomes.
- One upstream ESCI query is stored as the literal Python bytes representation `b'harrys razors for women'`. Every strategy received the same text, so paired comparisons remain valid, but the anomaly appears in the typed examples.
